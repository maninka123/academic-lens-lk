# Enriched merge audit
Merged 14 October 2026. Observations retain their source dates. One record per stable academic ID. Source and app coverage differ because the app already uses approved OpenAlex fallbacks.

```python
import json, csv
from pathlib import Path
root = Path.cwd()
while not (root / "Data files").exists(): root = root.parent
report = root / "reports/enriched-merge-2026-10-14"
summary = json.loads((report / "summary.json").read_text())
print({k: summary[k] for k in ["profiles_changed", "accepted_field_changes", "filled_missing_fields", "updated_existing_fields"]})
for field, counts in summary["coverage"].items():
    if counts["after"] != counts["before"]: print(field, counts, "gain", counts["after"] - counts["before"])
```

```python
from collections import Counter
with (report / "field-decisions.csv").open(encoding="utf-8-sig") as f: decisions = list(csv.DictReader(f))
print(Counter(d["decision"] for d in decisions))
accepted = [d for d in decisions if d["decision"] == "accepted"]
assert len({d["academic_id"] for d in accepted}) == summary["profiles_changed"]
assert len(accepted) == summary["accepted_field_changes"]
assert summary["app_after"]["academicRecords"] == summary["app_before"]["academicRecords"] == 1601
```
